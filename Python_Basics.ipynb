**Question 2** : Explain the difference between List and Tuple in Python.

**Ans -**


## Difference Between List and Tuple in Python

| Feature | List | Tuple |
|---|---|---|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutability | Changeable | Cannot be changed after creation |
| Speed | Generally slower | Generally faster |
| Use | When data needs to change | When data should remain fixed |


**Example:**

```python
# List
my_list = [10, 20, 30]
my_list[0] = 100
print(my_list)

# Tuple
my_tuple = (10, 20, 30)
print(my_tuple)
```

**Question 5** : Write a Python program to check whether a number is even or odd using conditional
statements.

**Ans -**

```python
# Take a number as input
num = int(input("Enter a number: "))

# Check whether the number is even or odd
if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

**Question 7** : Create a list of five numbers and print the maximum number from the list.

**Ans -**

```python
# Create a list of five numbers
numbers = [10, 25, 7, 45, 30]

# Find the maximum number
maximum = max(numbers)

# Print the maximum number
print("The maximum number is:", maximum)
```

**Question 8** : Write a Python program to remove duplicate values from a list using a set.

**Ans -**

```python
# Create a list with duplicate values
numbers = [10, 20, 10, 30, 20, 40, 30]

# Remove duplicates using a set
unique_numbers = set(numbers)

# Print the result
print("List after removing duplicates:", unique_numbers)
```

**Question 9** : Write a function that returns the square of a number.

**Ans -**

```python
# Define a function to calculate the square
def square(num):
    return num ** 2

# Call the function
result = square(5)

# Print the result
print("The square is:", result)
```

**Question 10** : Write a Python program to count how many times a number appears in a list.

Example List:

**[2,3,4,2,5,2]**

```python
# Example list
numbers = [2, 3, 4, 2, 5, 2]

# Number to count
num = 2

# Count how many times the number appears
count = numbers.count(num)

# Print the result
print("The number", num, "appears", count, "times.")
```
